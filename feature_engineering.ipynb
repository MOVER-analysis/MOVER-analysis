# Feature Engineering

```python
import pandas as pd
```

```python
df = pd.read_csv("data/cleaned_data.csv")

df.head()
```

### (1) Create the binary response variable from SMRTDTA_ELEM_VALUE

```python
df["hypoxemia"] = (
    df["SMRTDTA_ELEM_VALUE"]
    .str.lower()
    .str.contains("hypoxemia", na=False)
    .astype(int)
)

df.head()
```

### (2) Calculate BMI 

```python
df["bmi"] = df["WEIGHT"] / (df["HEIGHT"] ** 2)

df.head()
```

### (3) Convert SEX to binary (0 for male and 1 for female)

```python
df["sex"] = df["SEX"].map({"Male": 1, "Female": 0})
df.head()
```

```python
df.to_csv("data/feature_engineered_data.csv", index=False)
```
