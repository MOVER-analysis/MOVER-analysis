# Cohort Split

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("data/restructured_data.csv")
df.head()
```

```python
# filter all positives 
df_pos_id = df["hypoxemia"] == 1
df_pos = df[df_pos_id]
```

```python
# sample negatives
RANDOM_STATE = 42
NEG_SAMPLE_SIZE = 200

df_neg_id = df["hypoxemia"] == 0
df_neg = df[df_neg_id]

sample_neg = df_neg.sample(n=NEG_SAMPLE_SIZE, random_state=RANDOM_STATE)
```

```python
# combine the positives and negatives
sample = pd.concat([df_pos, sample_neg], axis=0)

sample.head(20)
```

```python
# train test (stratified) split
TEST_PROP = 0.25

train_df, test_df = train_test_split(
    sample,
    test_size = TEST_PROP,
    stratify = sample["hypoxemia"],
    random_state = RANDOM_STATE
)
```

```python
train_df.head()
```

```python
test_df.head()
```

```python
# outcome counts for train and test set

train_counts = train_df["hypoxemia"].value_counts()
test_counts = test_df["hypoxemia"].value_counts()

split_count = pd.DataFrame({
    "train": train_counts,
    "test": test_counts
}).T.fillna(0).astype(int)

split_count
```
